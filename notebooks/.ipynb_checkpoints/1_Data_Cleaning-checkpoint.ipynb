```python
import pandas as pd
import numpy as np
import os
import warnings
warnings.filterwarnings('ignore')
```

# 1. Load the Dataset

```python
df = pd.read_excel('../data/raw_data.xlsx',skiprows=1)
df.shape
```

# 2. Data Cleaning

```python
# Clean up column names by removing leading/trailing whitespace
df.columns = df.columns.str.strip()
```

```python
# Drop duplicate rows
initial_rows = len(df)
df.drop_duplicates(inplace=True)
print(f"Removed {initial_rows - len(df)} duplicate rows.")
print(df.shape)
```

### Drop columns that are empty

```python
# Find columns that contain only missing values
dropped_columns = df.columns[df.isna().all()].tolist()

# Drop them
df.dropna(axis=1, how='all', inplace=True)

print(f"Dropped {len(dropped_columns)} columns.")
print("Dropped columns:")
print(dropped_columns)

print(f"New shape: {df.shape}")
```

### Rename columns

```python
column_mapping = {
    'Heure': 'Hour',
    'Angle du vent(°)': 'Wind_Angle',
    'Vitesse vent(m/s)': 'Wind_Speed',
    'Humidité ambiante(%RH)': 'Ambient_Humidity',
    'Température ambiante(℃)': 'Ambient_Temperature',
    'Temp. (module PV)(℃)': 'PV_Temperature',
    'Irradiation journalière pente(Wh/㎡)': 'Daily_Irradiation_Cumulated',
    'Irradiation transitoire pente(W/㎡)': 'Transient_Irradiation',
    'Inverter1/Puissance active totale (kW)': 'AC',
    'Point de Rosée °C': 'Dewpoint',
    'Précipitation Totale mm': 'Rainfall',
    "Pression de l'air hPa": 'Air_Pressure'
}

df.rename(columns=column_mapping, inplace=True)
print("Renamed key columns for clarity.")
```

```python
df.columns
```

### Convert 'Hour' column to datetime objects

```python
df['Hour'] = pd.to_datetime(df['Hour'], errors='coerce')
```

### Remove rows before the meteostation started operating

```python
cutoff = pd.Timestamp('2024-05-18 11:00:00')
df = df[df['Hour'] >= cutoff].reset_index(drop=True)
print(f"Removed rows before {cutoff}. New shape: {df.shape}")
```

### Replace '--' with NumPy's Not a Number (NaN)

```python
# Replace placeholder characters ('--', etc.) with NumPy's Not a Number (NaN)
df.replace(['--', 'NA', 'NaN', 'nan'], np.nan, inplace=True)
print("Replaced placeholder text with NaN")
```

## Convert columns to numeric type

```python
cols_to_convert = [
    'Wind_Angle', 'Échelle vitesse vent()', 'Wind_Speed',
    'Ambient_Humidity', 'Ambient_Temperature', 'PV_Temperature',
    'Daily_Irradiation_Cumulated', 'Transient_Irradiation',
    'Radiation transitoire(W/㎡)', 'Inverter1/Puissance DC totale (kW)',
    'AC', 'pm10 μg/m³', 'pm2_5 μg/m³', 'carbon_monoxide μg/m³',
    'nitrogen_dioxide μg/m³', 'sulphur_dioxide μg/m³', 'ozone μg/m³',
    'aerosol_optical_depth Dimensionless', 'dust μg/m³', 'uv_index Index',
    'uv_index_clear_sky Index', 'ammonia μg/m³', 'alder_pollen Grains/m³',
    'birch_pollen Grains/m³', 'grass_pollen Grains/m³',
    'mugwort_pollen Grains/m³', 'olive_pollen Grains/m³',
    'ragweed_pollen Grains/m³', 'rain (mm)', 'dew_point_2m (°C)',
    'precipitation (mm)', 'cloud_cover (%)', 'cloud_cover_low (%)',
    'cloud_cover_mid (%)', 'cloud_cover_high (%)', 'wind_gusts_10m (km/h)',
    'wind_direction_100m (°)', 'wind_direction_10m (°)',
    'wind_speed_100m (km/h)', 'wind_speed_10m (km/h)'
]

missing = [c for c in cols_to_convert if c not in df.columns]
if missing:
    print("Columns not found in df (check):", missing)

for col in cols_to_convert:
    if col in df.columns:
        df[col] = (
            df[col]
            .astype(str)
            .str.replace(',', '.', regex=False)
            .str.strip()
        )
        df[col] = pd.to_numeric(df[col], errors='coerce')

print("Converted all feature columns to numeric type.")
```

## Compute Non-Cumulative Irradiation

The raw irradiation sensor reports a daily cumulative total that resets to zero at the start of each day. We decode this into a per-step (non-cumulative) value by differencing, treating any decrease as a daily reset.

```python
print(df['Daily_Irradiation_Cumulated'].head(30).to_list())
print(df['Daily_Irradiation_Cumulated'].dtype)
```

```python
# Calculate the difference between consecutive cumulative values
diff = df['Daily_Irradiation_Cumulated'].diff()

df['Daily_Irradiation'] = np.where(
    df['Daily_Irradiation_Cumulated'].isna(),
    np.nan,
    np.where(diff >= 0, diff, df['Daily_Irradiation_Cumulated'])
)

# Handle the first row
first_idx = df.index[0]
if pd.notna(df.loc[first_idx, 'Daily_Irradiation_Cumulated']):
    df.loc[first_idx, 'Daily_Irradiation'] = df.loc[first_idx, 'Daily_Irradiation_Cumulated']

df['Daily_Irradiation'] = df['Daily_Irradiation'].astype(float)

# Check the result
print(df[['Hour','Daily_Irradiation_Cumulated', 'Daily_Irradiation']].head(30))
```

## Save cleaned data

```python
df_cleaned = df
df_cleaned.to_csv("../data/cleaned_data.csv", index=False)
print(f"Saved cleaned data. Shape: {df_cleaned.shape}")
```
